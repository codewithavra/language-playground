````python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self._department = "IT"
        self.__salary = salary

emp = Employee("Alice", 75000)

# Accessing public attribute
print(emp.name)  # Output: Alice

# Accessing private attribute directly raises an AttributeError
# print(emp.__salary)  # Error!

# How Python handles private attributes (Name Mangling):
print(emp._Employee__salary)  # Output: 75000 (accessible, but discouraged)
````

```text
Modern Encapsulation using @property
Instead of writing explicit getter and setter methods like get_salary() and set_salary(), Python provides the @property decorator to manage access cleanly while using normal attribute syntax.
```

````python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance #Private Attribute
        
    # Getter: Exposes balance for reading
    @property
    def balance(self):
        return self.__balance

    # Setter: Validates data before modifying balance
    @balance.setter
    def balance(self, value):
        if value < 0:
            raise ValueError("Balance cannot be negative.")
        self.__balance = value
        
account = BankAccount("Bob", 500)

print(account.balance)
account.balance = -100
print(account.balance)
````
